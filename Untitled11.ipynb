```python
import pandas as pd
import numpy as np
import ast
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from scipy.spatial.distance import pdist, squareform


file_input = 'hasil_feature_engineering_umap.xlsx'

xls = pd.ExcelFile(file_input)
df_raw = pd.read_excel(xls, sheet_name='Data_Scraping')
df_prep = pd.read_excel(xls, sheet_name='Preprocessing')
df_w2v = pd.read_excel(xls, sheet_name='Word_Embedding')

corpus = df_prep['Teks_Gabungan'].tolist()
klub_list = df_raw['Nama Klub'].tolist()
n_klub = len(klub_list)


vec_tfidf = TfidfVectorizer()
X_tfidf = vec_tfidf.fit_transform(corpus)
sim_tfidf = cosine_similarity(X_tfidf)

vec_count = CountVectorizer()
X_count = vec_count.fit_transform(corpus)
sim_count = cosine_similarity(X_count)

vec_bin = CountVectorizer(binary=True)
X_bin = vec_bin.fit_transform(corpus).toarray()
sim_jaccard = 1 - squareform(pdist(X_bin, metric='jaccard'))

w2v_dict = {row['Kata']: row.drop('Kata').values.astype(float) for _, row in df_w2v.iterrows()}
tokens_per_klub = [ast.literal_eval(tok) if isinstance(tok, str) and tok.startswith('[') else tok for tok in df_prep['Token']]

doc_vectors = []
for tokens in tokens_per_klub:
    valid_vecs = [w2v_dict[w] for w in tokens if w in w2v_dict]
    doc_vectors.append(np.mean(valid_vecs, axis=0) if len(valid_vecs) > 0 else np.zeros(50))

sim_w2v = cosine_similarity(np.array(doc_vectors))

df_sim_tfidf = pd.DataFrame(sim_tfidf, index=klub_list, columns=klub_list)
df_sim_count = pd.DataFrame(sim_count, index=klub_list, columns=klub_list)
df_sim_jaccard = pd.DataFrame(sim_jaccard, index=klub_list, columns=klub_list)
df_sim_w2v = pd.DataFrame(sim_w2v, index=klub_list, columns=klub_list)

pairs = []
for i in range(n_klub):
    for j in range(i + 1, n_klub):
        pairs.append({
            'Klub 1': klub_list[i],
            'Klub 2': klub_list[j],
            'Cosine (TF-IDF)': sim_tfidf[i, j],
            'Cosine (Count)': sim_count[i, j],
            'Jaccard Similarity': sim_jaccard[i, j],
            'Cosine (Word2Vec)': sim_w2v[i, j]
        })

df_pairs_sorted = pd.DataFrame(pairs).sort_values(by='Cosine (TF-IDF)', ascending=False).reset_index(drop=True)


df_pairs_sorted.to_csv('hasil_text_similarity.csv', index=False)

with pd.ExcelWriter('hasil_text_similarity.xlsx', engine='openpyxl') as writer:
    df_pairs_sorted.to_excel(writer, sheet_name='Ranking_Similarity_Pasangan', index=False)
    df_sim_tfidf.to_excel(writer, sheet_name='Similarity_TFIDF')
    df_sim_count.to_excel(writer, sheet_name='Similarity_Count')
    df_sim_jaccard.to_excel(writer, sheet_name='Similarity_Jaccard')
    df_sim_w2v.to_excel(writer, sheet_name='Similarity_Word2Vec')

print("File Excel 'hasil_text_similarity.xlsx' & CSV berhasil dibuat!")

plt.figure(figsize=(12, 10))
sns.heatmap(df_sim_tfidf, annot=True, fmt=".2f", cmap='YlGnBu', cbar=True, linewidths=0.5)
plt.title('Heatmap Text Similarity (TF-IDF Cosine Similarity) - Klub Sepak Bola', fontsize=14, pad=15, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('heatmap_text_similarity.png', dpi=300)
plt.show()

plt.figure(figsize=(10, 6))
top10 = df_pairs_sorted.head(10).copy()
top10['Pasangan'] = top10['Klub 1'] + ' vs ' + top10['Klub 2']

bars = plt.barh(top10['Pasangan'][::-1], top10['Cosine (TF-IDF)'][::-1], color='#2b5c8f')
plt.title('Top 10 Pasangan Klub Paling Mirip (TF-IDF Cosine Similarity)', fontsize=13, pad=15, fontweight='bold')
plt.xlabel('Skor Kemiripan (Cosine Similarity)', fontsize=11)
plt.xlim(0, 0.6)

for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.008, bar.get_y() + bar.get_height()/2, f'{width:.4f}', va='center', ha='left', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('top10_similarity_barchart.png', dpi=300)
plt.show()
```
